# 1.
¿ Cuantos FTEs tiene la compañia en JULIO?
¿ que porcentaje de dedicación tiene los empleados durante el mes de Julio de 2019?

```python
#abro archivo con cambios aplicados 
df_plantilla_julio = pd.read_csv("../files/processed/plantilla_julio_2019.csv")
```

```python
# aunque la jornada en España es de 37.5 horas teóricas semanales  (código 156) y Brasil de 44 horas teoicas semanales (Código 185), 
# dado en la columna COD.UNIDAD DE NEGOCIO, COLUMNA Nº 6
```

```python
#Calculamos la plantilla total de julio 
headcount_julio = len(df_plantilla_julio)

print (headcount_julio)
```

```python
df_plantilla_julio.columns
```

```python

# Columnas clave
col_teoricas = "HORAS TEÓRICAS SEMANALES POR UNIDAD DE NEGOCIO"
col_trabajadas = "PROMEDIO DE HORAS  SEMANALES TRABAJADAS EN EL MES"



# 2. FTEs por empleado (respetando la jornada de España 37.5h y Brasil 44h)
df_plantilla_julio["FTE"] = df_plantilla_julio[col_trabajadas] / df_plantilla_julio[col_teoricas]
total_ftes = df_plantilla_julio["FTE"].sum()

# 3. Porcentaje de dedicación
total_horas_trabajadas = df_plantilla_julio[col_trabajadas].sum()
total_horas_teoricas = df_plantilla_julio[col_teoricas].sum()
porcentaje_dedicacion = (total_horas_trabajadas / total_horas_teoricas) * 100

print(f"Headcount Julio 2019: {headcount_julio:,}".replace(",", "."))
print(f"Total FTEs: {total_ftes:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))
print(f"Porcentaje de dedicación: {porcentaje_dedicacion:.2f}%".replace(".", ","))
```

4. Calcula la antigüedad media de los empleados que trabajan en Andalucía

```python
df_plantilla_julio.columns
```

```python
df_plantilla_julio['Localización Ciudad'].unique()
```

```python
# Con esto saco los trabajadores que son de Andalucia , los meto en una funcion

# 1. Lista de ciudades de Andalucía para filtrarlo por Andalucia
ciudades_andalucia = ['SEVILLA', 'MALAGA', 'GRANADA', 'CADIZ', 'HUELVA', 'ALMERIA', 'JAEN', 'CORDOBA']

# 2. Filtrar el DataFrame para quedarnos solo con Andalucía
df_andalucia = df_plantilla_julio[df_plantilla_julio['Localización Ciudad'].str.upper().isin(ciudades_andalucia)].copy()
```

```python
# 3. Calcular la antigüedad en años
df_andalucia['Fecha de Alta'] = pd.to_datetime(df_andalucia['Fecha de Alta'])
fecha_referencia = pd.to_datetime('2019-07-31')

# Antigüedad en días dividida entre 365.25 para tener en cuenta los años bisiestos
df_andalucia['Antiguedad_anios'] = (fecha_referencia - df_andalucia['Fecha de Alta']).dt.days / 365.25

# 4. Antigüedad media
antiguedad_media = df_andalucia['Antiguedad_anios'].mean()

print(f"Empleados en Andalucía: {len(df_andalucia)}")
print(f"Antigüedad media en Andalucía: {antiguedad_media:.2f} años".replace(".", ","))
```
